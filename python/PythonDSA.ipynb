```python
''' 
Problem: Find Minimum/Maximum in an Array
Constraint:
    1. The array is unsorted
    2. It can have both positive and negative number
Output:
    1. It should return the smallest element in the array
    2. It should return the largest element in the array
Edge Case:
    1. Empty array: return 'none'
    2. Array with one element: return same element for both min and max

'''
def findMinMax(arr):
    if len(arr) == 0:
        return None
    minimum = arr[0]
    maximum = arr[0]

    for num in arr:
        if num < minimum:
            minimum = num
        if num > maximum:
            maximum = num

    return {'Minimum': minimum, 'Maximum': maximum}

arr = [1000000, -1000000, 500000, -500000]
print(findMinMax(arr))
```

```python
'''
Problem: Check for Duplicates
Constraints:
    1. The array is unsorted
    2. The array can contain zero, positive and negative numbers
Output:
    1. Return true if the array contains duplicates else false.
Edge cases:
    1. Empty array: return false
    2. One element: return false

'''
from collections import defaultdict
def checkDuplicates(arr):
    if len(arr) <= 1:
        return False
    dupCount = defaultdict(int)

    for num in arr:
        dupCount[num] += 1
        if dupCount[num] > 1:
            return True

    return False

arr = [-10, 0, 5, -10, 20]
checkDuplicates(arr)
```

```python
'''
Problem: Move zeros to the end
Constraints:
    1. The solution should be done in place without making a copy of the array/list
Output:
    1. The array with all the zero elements should move to the last maintaining the relative order of the non-zero elements
Edge cases:
    1. Empty array: return the array as is
    2. Contains zeros only: return the array as is
    3. Contains no zero: return the array as is
'''
'''
Solution - 01: Using two pointer inplace arrangement
    1. This technique involves using two pointers to rearrange the elements in a collection without using extra space.
    2. Pointers initialization and role: 
        1. Left pointer: Start at the beginning of the array, it tracks the position where the next non zero element should go.
        2. Right pointer: Traverse the array, Looking for the elements that meets the certain criteria(In this case non-zero elements).
    3. Swapping/moving:
        1. When the element at the right pointer satisfies the condition(non-zero) it is swapped with the element at the right posistion.
        2. The left pointer moves forward to prepare for the next valid element
        3. The right pointer continues scanning the element.
'''

def moveZeros(arr):
    left = 0
    for right in range(1,len(arr)):
        if arr[left] == 0:
            temp = arr[left]
            if arr[right] != 0:
                arr[left] = arr[right]
                arr[right] = temp
                left +=1
        else:
            left+=1
    return arr

arr = [5, 0, 7, 8, 0, -3, 0, 2, 0]
moveZeros(arr)
```

```python
'''
Problem: Remove Duplicates from a Sorted Array
Constraints: 
    1. The solution should be done in-place, input array itself should be modified and no extra space should be created
    2. The relative order of the uniques element should be maintained
Output:
    1. Return the array removing the duplicate, keeping the relative order of the array.
Edge cases:
    1. Empty array: return array as is
    2. Array with one element: return the array as is
'''

def removeDuplicate(arr):
    left = 0
    for right in range(1,len(arr)):
        if arr[left] != arr[right]:
            left +=1
            arr[left] = arr[right]
    return arr[:left+1]

arr = [1, 1, 4,5,6,6,2, 2, 3, 3]
removeDuplicate(arr)
```

```python
'''
Problem: Partition Array into Even and Odd Numbers
Constraints:
    1. The operation must be performed in-place, meaning no additional space should be taken.
    2. The relative order of the even and odd numbers no need to be preserved.
Output:
    1. Rearrange the array such that all even numbers comes before the odd number
Edge cases:
    1. Empty array: return empty array
    2. Array with even num only: return array unchanged
    3. Array with odd num only: return array unchanged
    4. Array with one element: return array unchanged
'''

def partitionEvenOdd(arr):
    left = 0
    for right in range(1, len(arr)):
        if arr[left] % 2 != 0:
            temp = arr[left]
            if arr[right] % 2 ==0:
                arr[left] = arr[right]
                arr[right] = temp
                left+=1
        else:
            left += 1
    return arr

arr = [3, 1, 2, 4]
partitionEvenOdd(arr)
```

```python
'''
Problem: Segregate Positive and Negative Numbers
Constraints:
    1. The operation must be performed in-place, meaning no additional space to be taken.
    2. The relative order of the positive negative need not to be preserved.
Output:
    1. Rearrange the array such that all the negative numbers appears before positive numbers. Zero can be either treated as positive or negative.
Edge cases:
    1. Empty array: return an empty array
    2. All positive: return unchanged
    3. All negative: return unchanged
    4. Array with one element: return unchanged
    5. Array with zero: it can be either treated as positive or negative.
'''
def segregatePositiveNegative(arr):
    left = 0
    for right in range(1, len(arr)):
        if arr[left] >= 0:
            temp = arr[left]
            if arr[right] <0:
                arr[left] = arr[right]
                arr[right] = temp
                left +=1
        else:
            left +=1
    return arr

arr = [-1, -2, 3, 4, 5]
segregatePositiveNegative(arr)
```

```python
'''
Problem: Sort an Array of 0s, 1s, and 2s (Dutch National Flag Problem)
Constraints:
    1. The sorting must be done in place, meaning no extra space should be used
    2. The input array can be of any length including empty or single element array.
Output:
    1. Rearrange the array such that all 0's appears before 1's and 1's appear before 2.
Edge cases:
    1. Empty array: return as is
    2. Array with properly arranged: return as is
    3. Array with mixed arrangement: The array shoukd be properly arranged
    4. Array with one element: return as is
'''

def dutchNationalFlag(arr):
    left = 0
    mid =0
    right = len(arr)-1

    while mid <= right:
        leftTemp = arr[left]
        rightTemp = arr[right]
        if arr[mid] == 0:
            arr[left] = arr[mid]
            arr[mid] = leftTemp
            left += 1
            mid+=1
        elif arr[mid] == 1:
            mid += 1
        else:
            arr[right] = arr[mid]
            arr[mid] = rightTemp
            right -=1
    return arr

arr = [2,1,0,2,1]
dutchNationalFlag(arr)
```

```python
'''
Problem: Reverse a Subarray in Place
Constraints:
    1. The array can be of any leght empty, one element etc
    2. The rearrangement of the subarray needs to be done in-place meaning no additional extra space should be used
    3. The start and end indices are always valid and will be within the bounds
Output:
    1. The array with subarray between start and end (inclusive) reversed in-place
Edge cases:
    1. Empty array: return as is
    2. Array with one element: return as is
    3. start and end indices are same: return as is because there is no array to be reversed
    4. Full array reversal: start = 0 and end = len(arr)-1, the entire array should be reversed
    5. Partial subarray: Reverse only the portion of the array between start and end
'''
def reverseSubarray(arr, start, end):
    if len(arr) == 0:
        return arr

    while start<=end:
        temp = arr[start]
        arr[start] = arr[end]
        arr[end] = temp
        start += 1
        end -= 1
    return arr

arr = [7, 7, 7, 7, 7]
start = 1
end = 4
reverseSubarray(arr,start, end)
```

```python
'''
Problem: Longest substring without repeating characters
'''
from collections import defaultdict
def longestSubstring(str):
    start = 0
    count = 1
    seen  = defaultdict(int)

    for end in range(len(str)):
        seen[str[end]] +=1

        while seen[str[end]] > 1:
            seen[str[start]] -=1
            start +=1

        count = max(count, end-start+1)


    return count

print(longestSubstring("abcabcbb"))
```

```python
# Sorted array rotated at some point of time get the index in O(logn) time complexity

def rotatedBinarySearch(arr, k):
    left = 0
    right = len(arr)-1

    while left<=right:
        mid = (left + right) // 2
        if arr[mid] == k:
            return mid
        
        if arr[left] <= arr[mid]:
            if arr[left] <= k < arr[mid]:
                right = mid - 1
            else:
                left = mid + 1
        else:
            if arr[right] >= k > arr[mid]:
                left = mid +1
            else:
                right = mid -1
    return -1

arr = [4,5,6,1,2,3]
k = 5
print(rotatedBinarySearch(arr,k))
```

```python

def binarySearch(arr, k):
    left = 0
    right = len(arr) -1

    while left <= right:
        mid = (left+right)//2

        if arr[mid] == k:
            return mid
        
        if arr[mid] > k:
            right = mid - 1
        else:
            left = mid +1
            
    return -1

arr = [1,2,3,4,5]
k = 2
print(binarySearch(arr, k))
```
